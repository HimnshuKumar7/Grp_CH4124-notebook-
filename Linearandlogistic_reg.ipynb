```python
import numpy as np
```

```python
class LinearRegression:
  def __init__(self, lr= 0.01, epochs = 1000):
    self.lr = lr
    self.epochs = epochs

    self.W = 0
    self.b = 0

  def fit(self, X,y):
    m, n = X.shape
    self.W = np.zeros((n,1))
    self.b = 0.0
    y = y.reshape(-1,1)
    for epoch in range(self.epochs):
      y_pred = X @ self.W + self.b
      error = y_pred - y
      dw = (2/m) * (X.T @ error)
      db = (2/m) * (np.sum(error))
      self.W = self.W - self.lr*dw
      self.b = self.b - self.lr  * db

  def predict(self, X):
    return X @ self.W + self.b


```

```python
def sigmoid(z):
  return 1/(1+ np.exp(-z))

def derivative_sigmoid(z):
  return sigmoid(z)*(1-sigmoid(z))

class LogisticRegression:
  def __init__(self, lr= 0.01, epochs= 1000):
    self.lr  = lr
    self.epochs = epochs
    self.W = 0
    self.b = 0

  def fit(self, X,y):
    m,n = X.shape
    y = y.reshape((-1, 1))
    self.W = np.zeros((n,1))
    self.b = 0.0
    for epoch in range(self.epochs):
      z = X @ self.W + self.b
      y_pred = sigmoid(z)
      epsilon = 1e-15
      y_pred_clipped = np.clip(y_pred, epsilon, 1-epsilon)
      loss = -np.mean(y* np.log(y_pred_clipped) + (1-y)* np.log(1-y_pred_clipped))
      dw = (1/m) * (X.T @ (y_pred - y))
      db = (1/m) * np.sum(y_pred - y)
      self.W = self.W - self.lr * dw
      self.b = self.b - self.lr * db

    def predict_proab(self, X):
      z = X @ self.W + self.b
      return sigmoid(z)


    def predict(self, X):
      probability = self.predict_proab(X)
      labels = (probability >= 0.5).astype(int)
      return labels



```
